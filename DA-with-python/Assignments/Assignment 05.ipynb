```python
Create a list called numbers containing at least 8 integers.
```

```python
numbers = [10, 25, 30, 45, 50, 63, 70, 82]

print("All numbers:")
for number in numbers:
    print(number)

total = 0

for number in numbers:
    total = total + number

print("Sum =", total)

print("Even numbers:")

for number in numbers:
    if number % 2 == 0:
        print(number)
```

```python
Create a dictionary 
```

```python
student = {
    "name": "Ali",
    "age": 18,
    "course": "Computer Science",
    "marks": [65, 70, 80]
}

print("\nStudent Details:")

for key in student:
    print(key, ":", student[key])

total_marks = 0

for mark in student["marks"]:
    total_marks = total_marks + mark

average = total_marks / len(student["marks"])

print("Average marks =", average)

if average >= 50:
    print("Passed")
else:
    print("Failed")
```

```python
Create a list called employees.
```

```python
employees = [
    {"name": "Ali", "department": "IT", "salary": 50000},
    {"name": "Ahmed", "department": "HR", "salary": 45000},
    {"name": "Mehmood", "department": "Finance", "salary": 60000}
]

print("\nEmployee Details:")

for employee in employees:
    print("Name:", employee["name"])
    print("Department:", employee["department"])
    print("Salary:", employee["salary"])
    print()

highest = employees[0]
total_salary = 0

for employee in employees:
    total_salary = total_salary + employee["salary"]

    if employee["salary"] > highest["salary"]:
        highest = employee

print("Employee with highest salary:")
print(highest["name"])
print("Salary:", highest["salary"])

print("Total salary expense:", total_salary)
```

```python
While Loop with User Input
```

```python

numbers2 = []

while True:
    number = int(input("Enter a number (-1 to stop): "))

    if number == -1:
        break

    numbers2.append(number)

print("Complete list:", numbers2)

if len(numbers2) > 0:
    largest = numbers2[0]
    smallest = numbers2[0]

    for number in numbers2:
        if number > largest:
            largest = number

        if number < smallest:
            smallest = number

    print("Largest number:", largest)
    print("Smallest number:", smallest)
else:
    print("No numbers were entered.")
```

```python
Word frequency counter
```

```python
sentence = input("\nEnter a sentence: ")

words = sentence.split()

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] = frequency[word] + 1
    else:
        frequency[word] = 1

print("Word frequency:")

for word in frequency:
    print(word, ":", frequency[word])
```

```python
Nested Loops – Multiplication Table
```

```python
print("\nMultiplication Table:")

for i in range(1, 6):

    for j in range(1, 6):
        print(i * j, end=" ")

    print()
```
