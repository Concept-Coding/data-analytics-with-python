```python
numbers = input("Enter numbers separated by spaces: ").split()

print("Reversed list:")

for i in range(len(numbers) - 1, -1, -1):
    print(numbers[i])
```

```python
3. Find and display the largest number of a list without using built-in function max(). Your program should ask the user to input values in list from keyboard.
```

```python
numbers = input("Enter numbers separated by spaces: ").split()

largest = int(numbers[0])

for i in range(1, len(numbers)):
    if int(numbers[i]) > largest:
        largest = int(numbers[i])

print("Largest number is:", largest)
```

```python
4. Write a program that rotates the element of a list so that the element at the first index moves to the second index, the element in the second index moves to the third index, etc., and the element in the last index moves to the first index.
```

```python
numbers = input("Enter numbers separated by spaces: ").split()

last = numbers[-1]

for i in range(len(numbers) - 1, 0, -1):
    numbers[i] = numbers[i - 1]

numbers[0] = last

print("Rotated list:", numbers)
```

```python
6. Write a program that reads a string from the user containing a date in the form mm/dd/yyyy. It should print the date in the form March 12, 2021.
```

```python
date = input("Enter date in mm/dd/yyyy format: ")

parts = date.split("/")

month = int(parts[0])
day = parts[1]
year = parts[2]

months = ["January", "February", "March", "April", "May", "June",
          "July", "August", "September", "October", "November", "December"]

print(months[month - 1], day + ",", year)
```

```python
7. Write a program with a function that accepts a string from keyboard and create a new string after converting character of each word capitalized. For instance, if the sentence is "stop and smell the roses." the output should be "Stop And Smell The Roses".
```

```python
def capitalize_words(sentence):
    words = sentence.split()
    new_sentence = ""

    for word in words:
        new_sentence = new_sentence + word.capitalize() + " "

    return new_sentence


sentence = input("Enter a sentence: ")

result = capitalize_words(sentence)

print(result)
```

```python
Find the sum of each row of matrix of size m x n. For example for the following matrix output will be like this :2  11   7  125   2   9  158   3  10  42Sum of row 1 = 32Sum of row 2 = 31Sum of row 3 = 63
```

```python
rows = int(input("Enter number of rows: "))
columns = int(input("Enter number of columns: "))

matrix = []

for i in range(rows):
    row = []
    for j in range(columns):
        number = int(input("Enter number: "))
        row.append(number)
    matrix.append(row)

for i in range(rows):
    total = 0

    for j in range(columns):
        total = total + matrix[i][j]

    print("Sum of row", i + 1, "=", total)
```

```python
Write a program to add two matrices of size n x m.
```

```python
rows = int(input("Enter number of rows: "))
columns = int(input("Enter number of columns: "))

matrix1 = []
matrix2 = []

print("Enter values for first matrix:")

for i in range(rows):
    row = []
    for j in range(columns):
        number = int(input("Enter number: "))
        row.append(number)
    matrix1.append(row)

print("Enter values for second matrix:")

for i in range(rows):
    row = []
    for j in range(columns):
        number = int(input("Enter number: "))
        row.append(number)
    matrix2.append(row)

result = []

for i in range(rows):
    row = []

    for j in range(columns):
        row.append(matrix1[i][j] + matrix2[i][j])

    result.append(row)

print("Sum of the two matrices:")

for i in range(rows):
    print(result[i])
```

```python
Write a program to multiply two matrices
```

```python
rows1 = int(input("Enter rows of first matrix: "))
columns1 = int(input("Enter columns of first matrix: "))

rows2 = int(input("Enter rows of second matrix: "))
columns2 = int(input("Enter columns of second matrix: "))

matrix1 = []
matrix2 = []

print("Enter values for first matrix:")

for i in range(rows1):
    row = []
    for j in range(columns1):
        number = int(input("Enter number: "))
        row.append(number)
    matrix1.append(row)

print("Enter values for second matrix:")

for i in range(rows2):
    row = []
    for j in range(columns2):
        number = int(input("Enter number: "))
        row.append(number)
    matrix2.append(row)

result = []

for i in range(rows1):
    row = []

    for j in range(columns2):
        total = 0

        for k in range(columns1):
            total = total + matrix1[i][k] * matrix2[k][j]

        row.append(total)

    result.append(row)

print("Result of multiplication:")

for i in range(rows1):
    print(result[i])
```
