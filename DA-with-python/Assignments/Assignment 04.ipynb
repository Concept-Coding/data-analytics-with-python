Count how many times each number appears in a list using a dictionary.

```python
numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]

frequency = {}

for num in numbers:
    if num in frequency:
        frequency[num] = frequency[num] + 1  
    else:
        frequency[num] = 1                   
print(frequency)
```

Ask the user for a sentence, split it into words, and count each word's frequency.

```python
sentence = input("Enter a sentence: ")

words = sentence.split()

word_counts = {}

for word in words:
    if word in word_counts:
        word_counts[word] = word_counts[word] + 1
    else:
        word_counts[word] = 1

print(word_counts)
```

Calculate the total score and the average score from a dictionary of marks.

```python
marks = {"Math": 80, "English": 70, "Physics": 90}

total = 0
count = 0

for subject in marks:
    total = total + marks[subject]
    count = count + 1

average = total / count

print("Total Marks:", total)
print("Average Marks:", average)
```

Find which student has the highest score.

```python
students = {
    "Ali": 85,
    "Sara": 92,
    "Ahmed": 78,
    "Zara": 95
}

highest_score = 0
top_student = ""

for name in students:
    if students[name] > highest_score:
        highest_score = students[name]
        top_student = name

print("Highest scoring student is:", top_student, "with a score of", highest_score)
```

 Combine two separate dictionaries into one.

```python
dict1 = {"a": 1, "b": 2}
dict2 = {"c": 3, "d": 4}

merged_dict = dict1.copy()

for key in dict2:
    merged_dict[key] = dict2[key]

print(merged_dict)
```
