3. Write a program in which use all the operators we can use in Python

```python
# Arithmetic Operators
a = 10
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponent:", a ** b)

# Comparison Operators
print("Equal:", a == b)
print("Not Equal:", a != b)
print("Greater Than:", a > b)
print("Less Than:", a < b)
print("Greater Than or Equal:", a >= b)
print("Less Than or Equal:", a <= b)

# Assignment Operators
x = 10
x += 5
print("x += 5:", x)
x -= 2
print("x -= 2:", x)
x *= 2
print("x *= 2:", x)
x /= 2
print("x /= 2:", x)

# Logical Operators
p = True
q = False

print("AND:", p and q)
print("OR:", p or q)
print("NOT:", not p)

# Bitwise Operators
print("Bitwise AND:", a & b)
print("Bitwise OR:", a | b)
print("Bitwise XOR:", a ^ b)
print("Bitwise NOT:", ~a)
print("Left Shift:", a << 1)
print("Right Shift:", a >> 1)

```

4. Completes the following steps of small task:
    - Mention Marks of English , Islamiat and Maths out of 100 in 3 different variables
    - Mention Variable of Total Marks and assign 300 to it
    - Calculate Percentage

```python
english = 85
islamiat = 90
maths = 80

total_marks = 300

obtained_marks = english + islamiat + maths

percentage = (obtained_marks / total_marks) * 100

print("English Marks:", english)
print("Islamiat Marks:", islamiat)
print("Maths Marks:", maths)
print("Total Marks:", total_marks)
print("Obtained Marks:", obtained_marks)
print("Percentage:", percentage, "%")
```

5. Take the values of two variables `a` and `b`, then swap their values without using a third variable and print the result before and after swapping

```python
a = int(input("Enter value of a: "))
b = int(input("Enter value of b: "))

print("Before swapping:")
print("a =", a)
print("b =", b)

a, b = b, a

print("After swapping:")
print("a =", a)
print("b =", b)
```

6. Take the radius of a circle from a variable and calculate its Area and Circumference (use `pi = 3.14159`)

```python
radius = float(input("Enter the radius of the circle: "))

pi = 3.14159

area = pi * radius * radius
circumference = 2 * pi * radius

print("Area of the circle:", area)
print("Circumference of the circle:", circumference)
```

7. Take the price of an item and a discount percentage in variables, then calculate and print the discount amount and the final price after discount

```python
price = 2567
discount_percentage = 41

discount_amount = (price * discount_percentage) / 100
final_price = price - discount_amount

print("Original Price:", price)
print("Discount Amount:", discount_amount)
print("Final Price:", final_price)
```
